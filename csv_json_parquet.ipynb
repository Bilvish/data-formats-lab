```python
import json

order = {
    "order_id": 101,
    "customer": "Aman",
    "product": "Keyboard",
    "quantity": 1,
    "price": 1500
}

# Write Python data to a JSON file
with open("order.json", "w") as file:
    json.dump(order, file, indent=4)

print("JSON file created!")
```

```python

import json

order = {
    "order_id": 201,
    "customer": {
        "name": "Aman",
        "city": "Delhi"
    },
    "items": [
        {"product": "Keyboard", "quantity": 1, "price": 1500},
        {"product": "Mouse", "quantity": 2, "price": 500}
    ],
    "status": "Delivered"
}

# Save the nested dictionary as JSON
with open("nested_order.json", "w") as file:
    json.dump(order, file, indent=4)

# Read the JSON file
with open("nested_order.json", "r") as file:
    data = json.load(file)

print("Customer:", data["customer"]["name"])
print("City:", data["customer"]["city"])
print("First product:", data["items"][0]["product"])
print("Order status:", data["status"])
```

```python
total = sum(item["quantity"] * item["price"] for item in data["items"])
print("Order total:", total)
```

```python

import pandas as pd
import numpy as np

# Make results reproducible
np.random.seed(42)

n = 10_000

df = pd.DataFrame({
    "order_id": range(10001, 10001 + n),
    "customer_id": np.random.randint(1000, 3000, n),
    "product": np.random.choice(
        ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
        n
    ),
    "quantity": np.random.randint(1, 6, n),
    "unit_price": np.random.choice(
        [500, 1000, 1500, 5000, 25000, 55000],
        n
    ),
    "city": np.random.choice(
        ["Delhi", "Mumbai", "Pune", "Bengaluru", "Chennai"],
        n
    ),
    "status": np.random.choice(
        ["Delivered", "Shipped", "Pending", "Cancelled"],
        n
    )
})

df["total_amount"] = df["quantity"] * df["unit_price"]

print("Dataset created!")
print("Shape:", df.shape)
display(df.head())
```

```python
df.to_csv("orders.csv", index=False)

print("CSV saved successfully!")
```

```python
df.to_json("orders.json", orient="records", indent=2)

print("JSON saved successfully!")
```

```python
df.to_parquet("orders.parquet", index=False, engine="pyarrow")

print("Parquet saved successfully!")
```

```python
# Convert mismatched columns to original data types
for col in df.columns:
    df_csv[col] = df_csv[col].astype(df[col].dtype)
    df_json[col] = df_json[col].astype(df[col].dtype)

print("CSV matches:", df.equals(df_csv))
print("JSON matches:", df.equals(df_json))
print("Parquet matches:", df.equals(df_parquet))
```
